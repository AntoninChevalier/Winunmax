```python
#Initialisation des dataframes à partir des .csv bruts

df_clubs = pd.read_csv("../data/brute/clubs_fr.csv")
df_clubs.info()
display(df_clubs.head())
df_composition_equipes = pd.read_csv("../data/brute/game_lineups.csv")
df_composition_equipes.info()
display(df_composition_equipes.head())
df_detail_matchs = pd.read_csv("../data/brute/game_events.csv")
df_detail_matchs.info()
display(df_detail_matchs.head())
df_cartons_joueurs = pd.read_csv("../data/brute/player_appearance.csv")
df_cartons_joueurs.info()
display(df_cartons_joueurs.head())
df_valeur_joueurs = pd.read_csv("../data/brute/player_valuation_before_season.csv")
df_valeur_joueurs.info()
display(df_valeur_joueurs.head())
df_resultat_matchs = pd.read_csv("../data/brute/matchs_2013_2022.csv")
df_resultat_matchs.info()
display(df_resultat_matchs.head())
```

```python
#On remarque grâce aux .info() ci-dessus que les dates ne sont pas de type datetime

df_valeur_joueurs["date"] = pd.to_datetime(df_valeur_joueurs["date"])
df_composition_equipes["date"] = pd.to_datetime(df_composition_equipes["date"])
df_resultat_matchs["date"] = pd.to_datetime(df_resultat_matchs["date"])
df_cartons_joueurs["date"] = pd.to_datetime(df_cartons_joueurs["date"])
df_detail_matchs["date"] = pd.to_datetime(df_detail_matchs["date"])
```

```python
#Vérification si chaque équipe joue le même nombre de match : NON ---> on doit donc calculer le winrate

verification_match_domicile = df_resultat_matchs["home_club_id"].value_counts()
verification_match_exterieur = df_resultat_matchs["away_club_id"].value_counts()
verification_nb_total_match = verification_match_domicile.add(verification_match_exterieur,fill_value=0).astype(int)
verification_nb_total_match = verification_nb_total_match.reset_index()
verification_nb_total_match.head(5)
```

```python
#Test de la corrélation sur le dataframe de match initial

#Ajout de la différence de position, pour voir si la corrélation est meilleure avec le résultat ?
df_resultat_matchs["difference position"] = df_resultat_matchs["away_club_position"]-df_resultat_matchs["home_club_position"]

nettoyage.afficher_matrice_correlation(df_resultat_matchs)
```

```python

#Calcul de la valorisation par équipe, ici seulement les 11 de départs

#On va trier d’abord par player_id puis par date, mais également enlever les remplaçants du calcul de coût
df_compo_triee = df_composition_equipes.sort_values(by=['player_id']).sort_values(by=["date"])
df_compo_triee = df_compo_triee[df_compo_triee["type"] == "starting_lineup"]
df_valeurs_triee = df_valeur_joueurs.sort_values(by=['player_id']).sort_values(by=["date"])

df_compo_valeur = pd.merge_asof(df_compo_triee,df_valeurs_triee,on='date',by='player_id',direction='backward')
df_valeur_equipes = df_compo_valeur.groupby(['game_id', 'club_id'])['market_value_in_eur'].sum().reset_index()
df_valeur_equipes.rename(columns={'market_value_in_eur': 'team_market_value'}, inplace=True)

df_valeur_equipes.head(8)
```
